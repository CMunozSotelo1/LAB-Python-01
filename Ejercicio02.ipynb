```python
n = int(input("Ingrese un número: "))
suma = 0
for i in range(1, n + 1):
    suma += i

    if i % 2 == 0:
        print("Número par:", i)
print("La suma de los primeros", n, "números naturales es:", suma)
```

Ejercicio 3

```python
temperaturas = [20, 25, 18, 30, 22]
suma = 0
for temperatura in temperaturas:
    suma += temperatura
promedio = suma / len(temperaturas)
print("Temperaturas:", temperaturas)
print("Promedio:", promedio)
print("Temperaturas mayores al promedio:")
for temperatura in temperaturas:
    if temperatura > promedio:
        print(temperatura)
```

```python
suma_pares = 0
suma_impares = 0

for i in range(1, 11):

    if i % 2 == 0:
        suma_pares += i
    else:
        suma_impares += i

estado_final = {
    "suma_pares": suma_pares,
    "suma_impares": suma_impares
}

print("Estado final:", estado_final)
```

Ejercicio 6

```python
productos = []
while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre == "fin":
        break
    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad: "))
    if precio < 0 or cantidad < 0:
        print("Producto inválido. Intente nuevamente.")
        continue
    productos.append({
        "nombre": nombre,
        "precio": precio,
        "cantidad": cantidad
    })
total = 0
for p in productos:
    total += p["precio"] * p["cantidad"]
print("Total del inventario:", total)
if len(productos) > 0:
    producto_mas_caro = max(productos, key=lambda p: p["precio"])
    print("Producto más caro:", producto_mas_caro)
else:
    print("No hay productos registrados")

print("Estado final del inventario:")
print(productos)
```

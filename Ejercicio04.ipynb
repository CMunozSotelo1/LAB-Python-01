1. Variables globales y funciones
A. Ejecutar el código original. ¿Qué error se obtiene y por qué ocurre?

```python
contador_global = 0

def incrementar():
    contador_global = contador_global + 1

incrementar()
print(contador_global)
```

```python
UnboundLocalError: cannot access local variable 'contador_global'
where it is not associated with a value
```

El error ocurre porque Python interpreta contador_global como una variable local al encontrar una asignación dentro de la función. Sin embargo, intenta utilizarla antes de asignarle un valor local.

B. Primera solución: utilizando global

```python
contador_global = 0
def incrementar():
    global contador_global
    contador_global = contador_global + 1
incrementar()
print(contador_global)
```

C. Segunda solución: recibir el valor y retornar el nuevo valor

```python
def incrementar(contador):
    contador = contador + 1
    return contador
contador_global = 0
contador_global = incrementar(contador_global)
print(contador_global)
```

D. ¿Cuál de las dos versiones es más fácil de probar y por qué?


La segunda versión es más fácil de probar porque recibe un valor y devuelve un resultado sin modificar directamente variables globales. Esto permite probar la función de manera independiente con diferentes valores y facilita encontrar errores.

2. Listas, objetos mutables y función id()

A. Antes de ejecutar el código, predecir qué se imprime y comprobarlo

```python
def agregar_nota(notas, nueva):
    notas.append(nueva)
def reiniciar(notas):
    notas = []
mis_notas = [3.5, 4.0]
agregar_nota(mis_notas, 4.8)
reiniciar(mis_notas)
print(mis_notas)
#Predicción: Se imprime [3.5, 4.0, 4.8].
```

B. Explicar la diferencia entre modificar un objeto mutable y reasignar una variable local

- Modificar con append(): agrega un elemento directamente a la lista original, conservando el mismo objeto.
- Reasignar con notas = []: crea una lista nueva y hace que la variable local apunte a ella. La lista original permanece igual.

C. Modificar reiniciar para vaciar la lista original sin utilizar return

```python
def agregar_nota(notas, nueva):
    notas.append(nueva)
def reiniciar(notas):
    notas.clear()
mis_notas = [3.5, 4.0]
agregar_nota(mis_notas, 4.8)
reiniciar(mis_notas)
print(mis_notas)
```

D. Utilizar id() para mostrar cuándo la variable apunta al mismo objeto y cuándo no

```python
def agregar_nota(notas, nueva):
    print("ID dentro de agregar:", id(notas))
    notas.append(nueva)
def reiniciar(notas):
    print("ID antes de limpiar:", id(notas))
    notas.clear()
    print("ID después de limpiar:", id(notas))
mis_notas = [3.5, 4.0]
print("ID original:", id(mis_notas))
agregar_nota(mis_notas, 4.8)
print("Lista después de agregar:", mis_notas)
reiniciar(mis_notas)
print("Lista después de reiniciar:", mis_notas)
print("ID final:", id(mis_notas))
```

3. Cajero automático con funciones

```python
cuenta = {
    "titular": "Ana",
    "saldo": 200_000,
    "movimientos": []
}
def consultar(cuenta):
    print("\nTitular:", cuenta["titular"])
    print("Saldo actual:", cuenta["saldo"])
def depositar(cuenta, monto):
    if monto <= 0:
        print("El monto debe ser mayor que cero")
    else:
        cuenta["saldo"] += monto
        movimiento = {
            "tipo": "Deposito",
            "monto": monto,
            "saldo": cuenta["saldo"]
        }
        cuenta["movimientos"].append(movimiento)
        print("Deposito realizado correctamente")
def retirar(cuenta, monto):
    if monto <= 0:
        print("El monto debe ser mayor que cero")
    elif monto > cuenta["saldo"]:
        print("Saldo insuficiente")
    else:
        cuenta["saldo"] -= monto
        movimiento = {
            "tipo": "Retiro",
            "monto": monto,
            "saldo": cuenta["saldo"]
        }
        cuenta["movimientos"].append(movimiento)
        print("Retiro realizado correctamente")
while True:
    print("\n===== CAJERO AUTOMATICO =====")
    print("1. Consultar saldo")
    print("2. Depositar dinero")
    print("3. Retirar dinero")
    print("4. Salir")
    opcion = input("Seleccione una opcion: ")
    if opcion == "1":
        consultar(cuenta)
    elif opcion == "2":
        monto = float(input("Ingrese el monto a depositar: "))
        depositar(cuenta, monto)
    elif opcion == "3":
        monto = float(input("Ingrese el monto a retirar: "))
        retirar(cuenta, monto)
    elif opcion == "4":
        print("\nSaliendo del cajero...")
        break
    else:
        print("Opcion no valida")
print("\n===== EXTRACTO BANCARIO =====")
print("Titular:", cuenta["titular"])
print("Saldo final:", cuenta["saldo"])
print("-" * 55)
print(f"{'TIPO':<15}{'MONTO':<15}{'SALDO':<15}")
print("-" * 55)
for movimiento in cuenta["movimientos"]:
    print(
        f"{movimiento['tipo']:<15}"
        f"{movimiento['monto']:<15.2f}"
        f"{movimiento['saldo']:<15.2f}"
    )
print("-" * 55)
print("Fin del extracto")
```

```python
saldo = float(input("Ingrese el saldo inicial: "))
deposito = float(input("Ingrese el monto a depositar: "))
retiro = float(input("Ingrese el monto a retirar: "))
interes = float(input("Ingrese el interés mensual (ej. 0.02): "))

saldo = saldo + deposito
saldo = saldo - retiro

saldo_final = saldo * (1 + interes)

print(f"Saldo final con interés: ${saldo_final:.2f}")
```

¿Qué representan los estados?
- Estado inicial: Acá estan los valores que tienen las variables antes de pasar por las operaciones que le apliquemos.


- Estado final: Los valores finales ya despues de haber aplicado todas las operaciones del codigo.


¿Cómo afecta cada instrucción a la memoria?


- saldo = 500_000: crea la variable saldo y almacena el valor 500000.
- deposito = 150_000: almacena el monto del depósito.
- saldo = saldo + deposito: calcula la suma y actualiza el valor de saldo.
- saldo = saldo - retiro: resta el retiro y vuelve a actualizar saldo.
- saldo_final = saldo * (1 + interes): calcula el saldo con intereses y lo almacena en una nueva variable.
- print(): muestra los valores en pantalla, sin modificar las variables.

```python
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]

temperaturas.append(23.7)
temperaturas.pop(0)
temperaturas.insert(1, 22.0)

print("Temperaturas:", temperaturas)

# Ordenar de menor a mayor
temperaturas.sort()
print("Ordenadas:", temperaturas)

# Diferencia entre máxima y mínima
diferencia = max(temperaturas) - min(temperaturas)
print("Diferencia:", diferencia)

# Promedio usando un ciclo
suma = 0

for temperatura in temperaturas:
    suma = suma + temperatura

promedio = suma / len(temperaturas)
print("Promedio:", promedio)

# Buscar 23.7
indice = -1

for i in range(len(temperaturas)):
    if temperaturas[i] == 23.7:
        indice = i
        break

print("Índice de 23.7:", indice)

# Segunda a cuarta posición
print("Posiciones 2 a 4:", temperaturas[1:4])

# Tres primeras temperaturas
print("Tres primeras:", temperaturas[:3])
```

```python
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10}
]

# Nombre del primer producto
print("Primer producto:", inventario[0]["nombre"])

# Agregar producto
nuevo_producto = {
    "nombre": "Arroz",
    "precio": 2500,
    "stock": 8
}
inventario.append(nuevo_producto)

# Vender dos unidades de leche
for producto in inventario:
    if producto["nombre"] == "Leche":
        producto["stock"] = producto["stock"] - 2
        break

# Valor total del inventario
valor_total = 0

for producto in inventario:
    valor_total += producto["precio"] * producto["stock"]

print("Valor total:", valor_total)

# Productos con stock menor a 10
print("Stock menor a 10:")

for producto in inventario:
    if producto["stock"] < 10:
        print(producto)

# Aumentar precios en 10 %
for producto in inventario:
    producto["precio"] = round(producto["precio"] * 1.10, 2)

# Buscar producto
nombre_buscar = input("Nombre del producto a buscar: ")
encontrado = False

for producto in inventario:
    if producto["nombre"].lower() == nombre_buscar.lower():
        print("Producto encontrado:", producto)
        encontrado = True
        break

if not encontrado:
    print("Producto no encontrado")

# Eliminar productos con stock cero
i = 0

while i < len(inventario):
    if inventario[i]["stock"] == 0:
        inventario.pop(i)
    else:
        i += 1

# Ordenar por precio ascendente
for i in range(len(inventario)):
    for j in range(i + 1, len(inventario)):
        if inventario[i]["precio"] > inventario[j]["precio"]:
            temporal = inventario[i]
            inventario[i] = inventario[j]
            inventario[j] = temporal

print("Inventario final:", inventario)
```

```python
punto = (3, 7)
print("Punto:", punto)

ancho = float(input("Ingrese el ancho: "))
alto = float(input("Ingrese el alto: "))

rectangulo = (ancho, alto)

area = rectangulo[0] * rectangulo[1]
perimetro = 2 * (rectangulo[0] + rectangulo[1])

print("Área:", area)
print("Perímetro:", perimetro)

# Intentar cambiar el ancho
try:
    rectangulo[0] = 10
except TypeError:
    print("Error: las tuplas son inmutables")

# Distancia total recorrida
posiciones = [(0, 0), (1, 2), (3, 4)]
distancia_total = 0

for i in range(len(posiciones) - 1):
    x1 = posiciones[i][0]
    y1 = posiciones[i][1]

    x2 = posiciones[i + 1][0]
    y2 = posiciones[i + 1][1]

    distancia = ((x2 - x1)**2 + (y2 - y1)**2)**0.5
    distancia_total = distancia_total + distancia

print("Distancia total:", distancia_total)
```

```python
biblioteca = {
    "libros": [
        {"titulo": "1984", "autor": "George Orwell", "prestado": False},
        {"titulo": "El Principito",
         "autor": "Antoine de Saint-Exupéry",
         "prestado": True}
    ],
    "usuarios": ["Ana", "Luis", "María"]
}

# Contar libros prestados y disponibles
prestados = 0
disponibles = 0

for libro in biblioteca["libros"]:
    if libro["prestado"]:
        prestados += 1
    else:
        disponibles += 1

print("Libros prestados:", prestados)
print("Libros disponibles:", disponibles)

# Agregar un libro
nuevo_libro = {
    "titulo": "Cien años de soledad",
    "autor": "Gabriel García Márquez",
    "prestado": False
}

biblioteca["libros"].append(nuevo_libro)

# Marcar 1984 como prestado
for libro in biblioteca["libros"]:
    if libro["titulo"] == "1984":
        libro["prestado"] = True
        break

# Mostrar estado actualizado
print(biblioteca)
```
